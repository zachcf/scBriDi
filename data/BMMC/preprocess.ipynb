The two input `.h5ad` files are downloaded from [GEO: GSE194122](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE194122).

```python
import sys
sys.path.append('../..')
import scanpy as sc
from src.processing import ADT_data_preprocessing, RNA_data_preprocessing, ATAC_data_preprocessing
```

```python
multiome_data = sc.read_h5ad('GSE194122_openproblems_neurips2021_multiome_BMMC_processed.h5ad')
cite_data = sc.read_h5ad('GSE194122_openproblems_neurips2021_cite_BMMC_processed.h5ad')
multiome_data.X = multiome_data.layers["counts"]
cite_data.X = cite_data.layers["counts"]
```

```python
gex_multiome = multiome_data[:, multiome_data.var['feature_types'] == 'GEX']
atac_multiome = multiome_data[:, multiome_data.var['feature_types'] == 'ATAC']

gex_cite = cite_data[:, cite_data.var['feature_types'] == 'GEX']
adt_cite = cite_data[:, cite_data.var['feature_types'] == 'ADT']
```

```python
adt_cite = ADT_data_preprocessing(adt_cite)[0]
atac_multiome = ATAC_data_preprocessing(atac_multiome)[0]
atac_multiome.write('ATAC_1.h5ad')
adt_cite.write('ADT_2.h5ad')
```

```python
gex_processed_1 = RNA_data_preprocessing(gex_multiome, n_top_genes=3000)
gex_processed_2 = RNA_data_preprocessing(gex_cite, n_top_genes=3000)

common_genes = set(gex_processed_1.var_names) & set(gex_processed_2.var_names)
gex_common_1 = gex_processed_1[:, gex_processed_1.var_names.isin(common_genes)]
gex_common_2 = gex_processed_2[:, gex_processed_2.var_names.isin(common_genes)]

gex_common_1.write('GEX_c1.h5ad')
gex_common_2.write('GEX_c2.h5ad')
```
